## Figure 1

Three aligned rows: titles, sketches, plots.

- **Left, an intervention.** The same individuals, models and prompts with reasoning off (Qwen off,
  OpenAI none) and on (Qwen on, OpenAI high). Top: one light line per model pair (36), the medians
  on top; the pair values are those of `fig:agree-effort`. Bottom, on the same off / on axis: the
  share of individuals all nine models get wrong (the 0-correct level of `fig:effort-recourse`), as
  bars from zero.
- **Right, observed behavior.** Reasoning on; individuals *sorted* by the length they drew, so
  different x holds different individuals and nobody's budget is raised. The pooled curve of
  `fig:agree-length`: per bin, the MEAN over model pairs of kappa recomputed within the bin (the left
  plot uses the median; both are labelled). Faint lines are single seed draws. Below: how the
  individuals of the longest bin split into all nine correct / mixed / all nine wrong.

The two kappa plots have the same limits, height and position, so equal differences have equal
lengths on the page. The sketches are illustrative. Saved as PDF/PNG (LaTeX text), plus an SVG and
an `-editable.pdf` with plain text in the bundle's non-LaTeX fonts.
